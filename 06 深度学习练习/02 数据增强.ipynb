```python
from torchvision import transforms
from PIL import Image
import matplotlib.pyplot as plt

image_path = r'E:\03学习资料\coding_class\04 深度学习\25.02.15-猫狗二分类任务\dataset\training_set\cats\cat.1.jpg'
image = Image.open(image_path)

trans = [
    # 以概率p对图像进行水平翻转
    transforms.RandomHorizontalFlip(p=0.5),
    # 在[-degress,degress]之间随机旋转
    transforms.RandomRotation(degrees=45),
    # 随机切剪图像，然后缩放到指定尺寸
    # size:缩放后的尺寸
    # scale:裁剪的面积范围
    transforms.RandomResizedCrop(size=224, scale=(0.2,0.5)),
    # 随机改变图像的亮度、对比度、饱和度与色调
    # 范围：[max(0,1-brightness),1+brightness]
    transforms.ColorJitter(brightness=0.4, contrast=0.4, saturation=0.4,hue=0.1),
    # 图像保持中心不变，随机放射变换
    # degrees:旋转的角度范围
    # translate:平移的比例范围
    # scale：缩放的范围
    # shear: 裁剪的角度范围，效果：图像会沿着一个方向发生倾斜
    transforms.RandomAffine(degrees=45, translate=(0.1,0.1), scale=(0.8, 1.2), shear=10)
]

fig,ax = plt.subplots( 5, 5, figsize=(15,20))
for idx, t in enumerate(trans):
    for i in range(5):
        ax[idx,i].imshow(t(image))
        ax[idx,i].axis('off')

plt.tight_layout()
```
