```python
import os


resize_path = r"E:\Desktop\data\resize_image"

def rename_images(resize_path):
    for filename in os.listdir(resize_path):
        image_path = os.path.join(resize_path, filename)

        name, ext = os.path.splitext(filename)
        new_name = f"{name}_strat{ext}"
        new_image_path = os.path.join(resize_path, new_name)
        os.rename(image_path, new_image_path)
        print(f"Renamed: {filename} -> {new_name}")

rename_images(resize_path)
```
