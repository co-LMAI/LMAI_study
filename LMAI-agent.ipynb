```python
import os
from dotenv import load_dotenv
load_dotenv()
from huggingface_hub import login


token = os.getenv('HF_TOKEN')
login(token=token)
```

```python
from smolagents import InferenceClientModel
model = InferenceClientModel()

messages = [{"role": "user", "content": "将所有的输出内容都替换为中文."}]
response = model(messages, stop_sequences=["END"])
response
```

```python
from smolagents import CodeAgent, DuckDuckGoSearchTool, InferenceClientModel


# codeagent:是smolagents中，快速构件agent的默认方式，使用代码与大模型交互
# DuckDuckGoSearchTool：一个网页搜索工具
agent = CodeAgent(tools=[DuckDuckGoSearchTool()], model=InferenceClientModel())

agent.run("给我一个施工图做法")
```

```python
from smolagents import tool
import requests

url = 'http://127.0.0.1:11434/api/chat'

payload = {
    "model": "qwen2",
    "messages": [
        {"role": "system", "content": "你正在面对一个中文人员，对话中你要保证所有内容都输出中文"},
        {"role": "user", "content": "给我一个排水沟做法?"}
    ],
    "stream": False
}

res = requests.post(url, json=payload)
print(res.json()["message"]["content"])
```

```python
from smolagents import CodeAgent, DuckDuckGoSearchTool,LiteLLMModel

model = LiteLLMModel(
    model_id="ollama/qwen2:7b", # 格式是ollama_chat/xxx
    base_url="http://127.0.0.1:11434",
    api_key='ollama', # 默认为 None
)


agent = CodeAgent(tools=[DuckDuckGoSearchTool()], model=model)

agent.run("给我一个施工图做法")
```
